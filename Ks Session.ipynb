```python
from tkinter import *
from tkinter import messagebox
import pandas as pd
import os

# Define the path to the Excel file using raw string to avoid Unicode error
excel_file = r'C:\Users\aloga\OneDrive\Documents\Book1.xlsx'

def submit_form():
    try:
        name = name_entry.get()
        age = age_entry.get()
        email = email_entry.get()
        
        # Validate form data
        if not name or not age or not email:
            messagebox.showerror("Eror","All fields are required!")
            return
        
        # Create a DataFrame with the form data
        data = {'Name': [name], 'Age': [age], 'Email': [email]}
        df = pd.DataFrame(data)
        
        # Check if the Excel file exists
        if not os.path.exists(excel_file):
            # If it doesn't exist, create it and write the data
            df.to_excel(excel_file, index=False)
        else:
            # If it exists, append the data without writing the header
            with pd.ExcelWriter(excel_file, mode='a', engine='openpyxl', if_sheet_exists='overlay') as writer:
                df.to_excel(writer, index=False, header=False, startrow=writer.sheets['Sheet1'].max_row)
        
        # Show a confirmation message
        messagebox.showinfo("Form Submited", "Your details have been added to the Excel sheet.")
        
        #Clear the form fields
        name_entry.delete(0, END)
        age_entry.delete(0, END)
        email_entry.delete(0, END)

    except Exception as e:
        messagebox.showerror("Error", f"An error occurred: {e}")
        print(f"An error occurred: {e}")

# Initialize the Tkinter window
root = Tk()
root.title("Entry Form")
root.geometry("500x600")

# Create a frame for the form
frame = Frame(root, padx=10, pady=10)
frame.pack(pady=20)

# Create form fields
Label(frame, text="Name:").grid(row=0, column=0, sticky=W)
name_entry = Entry(frame, width=30)
name_entry.grid(row=0, column=1)

Label(frame, text="Age:").grid(row=1, column=0, sticky=W)
age_entry = Entry(frame, width=30)
age_entry.grid(row=1, column=1)

Label(frame, text="Email:").grid(row=2, column=0, sticky=W)
email_entry = Entry(frame, width=30)
email_entry.grid(row=2, column=1)

# Add a submit button
submit_button = Button(frame, text="Submit", command=submit_form)
submit_button.grid(row=3, columnspan=2, pady=10)

# Run the application
root.mainloop()
```
